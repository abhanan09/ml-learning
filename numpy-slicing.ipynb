```python
import numpy as np

a = np.arange(1,26).reshape(5, 5)
print(a)
```

```python
#exercise 1 (The first row)

first_row = a[0 ,:]
print(first_row)
```

```python
#exercise 2 (The last row, use a negative index)

last_row = a[-1 ,:]
print(last_row)
```

```python
#exercise 3 (The third column)

third_column = a[: , 2]
print(third_column)
```

```python
#exercise 4 (The first two rows)

first_two_rows = a[0:2 , :]
print(first_two_rows)
```

```python
#exercise 5 (The last two columns)

last_two_columns = a[: , 3:5]
print(last_two_columns)
```

```python
#exercise 6 (The top-left 3x3 block)

ex_6 = a[0:3 , 0:3]
print(ex_6)
```

```python
#exercise 7 (The bottom-right 2x2 block)

ex_7 = a[3:5 , 3:5]
print(ex_7)
```

```python
#exercise 8 (Every second row (step))

every_second_row = a[0:5:2]
print(every_second_row)
```

```python
#exercise 9 (Every second column of the first three rows)

ex_9 = a[0:3 ,::2]
print(ex_9)
```

```python
#exercise 10 (The center element (row 3, column 3, counting from 1))

center_element = a[2,2]
print(center_element)
```

```python
#exercise 11 (The array with rows reversed)

reversed_rows = a[::-1 , :]
print(reversed_rows)
```

```python
#exercise 12 (The outer border as one set: the first and last row, and the first and last column)

top = a[0,0:5]
bottom = a[4,0:5]
left = a[1:4,0]
right = a[1:4,4]

border = np.concatenate([top,bottom,left,right])
print(border)

```
